## 2. Eksperimen DeFi: Simulasi APR vs APY & Impermanent Loss 🧑‍💻
Bagian ini menjalankan simulasi matematis efek pemajemukan (*compounding*) serta risiko *Impermanent Loss* pada AMM Constant Product ($x \cdot y = k$).

```python
import math

def hitung_apy(apr_persen, n_periode=365):
    """Menghitung APY dari APR dengan frekuensi compounding n."""
    r = apr_persen / 100.0
    return (((1.0 + (r / n_periode)) ** n_periode) - 1.0) * 100.0

def hitung_impermanent_loss(rasio_harga):
    """Menghitung Impermanent Loss (%) pada AMM Constant Product."""
    if rasio_harga <= 0:
        raise ValueError("Rasio harus positif")
    return ((2.0 * math.sqrt(rasio_harga) / (1.0 + rasio_harga)) - 1.0) * 100.0

# 🧑‍💻 Variabel Eksperimen
apr_input = 24.0  # APR 24%
apy_harian = hitung_apy(apr_input, n_periode=365)
print(f"[Simulasi APR ke APY] APR: {apr_input:.2f}% -> APY Daily: {apy_harian:.2f}%")

print("\n[Simulasi Impermanent Loss]")
skenario_k = [1.25, 1.50, 2.00, 3.00, 5.00, 0.50]
for k in skenario_k:
    il = hitung_impermanent_loss(k)
    print(f" - Rasio Harga {k*100:3.0f}%: Impermanent Loss = {il:6.2f}%")
```

### Analisis Hasil ✍️
- **Efek Pemajemukan:** Dengan bunga nominal APR 24%, auto-compound harian menghasilkan APY riil sebesar 27.11% (+3.11%).
- **Impermanent Loss:** Jika harga salah satu token naik 2x (200%), LP menderita kerugian -5.72% dibanding menyimpan token di dompet (HODL). Jika melonjak 5x, kerugian mencapai -25.46%.
