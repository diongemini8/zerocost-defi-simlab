## 1. Studi Kasus & Analisis "Dari Mana Uangnya Berasal?" ✍️
Menelusuri arus likuiditas dan nilai ekonomi riil pada ekosistem DeFi:
1. **Lending Protocol (Aave):** Bunga berasal dari peminjam (*borrowers*) yang membayar *interest rate* untuk lindung nilai atau likuiditas operasional.
2. **AMM DEX (Uniswap):** Imbal hasil berasal dari *swap fee* transaksi riil pengguna (0.3% per swap) yang dibagikan kepada Liquidity Provider.
3. **Proof-of-Stake:** Reward berasal dari pencetakan koin inflasi protokol terprogram dan tip transaksi blok validator.
4. **Yield Farming:** Subsidi token tata kelola protokol; waspada terhadap risiko dilusi harga jika protokol tidak memiliki *real protocol revenue*.
